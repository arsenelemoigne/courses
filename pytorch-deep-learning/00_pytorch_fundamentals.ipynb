Matrix multiplication

Two main ways:
- element wise mutliplication
- matrix multiplication (= dot product)

https://www.mathsisfun.com/algebra/matrix-multiplying.html$0

1. The inner dimensions must match :

- (3, 2) @ (3, 2) will not work
- (2, 3) @ (3, 2) will work
- (3, 2) @ (2, 3) will work

2. The resulting matrix has the shape of the outer dimension
- (2, 3) @ (3, 2) -> (2, 2)
- (3, 2) @ (2, 3) -> (3, 3)




```python
# Element wise multiplication

tensor = torch.tensor([1, 2, 3])

print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# Matric multiplication

torch.matmul(tensor, tensor)      # this is the correct PyTorch method to multiply using : 'torch.matmul(a, b)'

# 1 x 1 + 2 x 2 + 3 x 3 = 14
```

```python
%%time
value = 0
for i in range(len(tensor)):
  value +=tensor[i] * tensor[i]
print(value)
```

```python
%%time
torch.matmul(tensor, tensor)

# ten times quicker compared to the for loop
# use torch operations if available
```

Rashaping, stacking, squeezing and unsqueezing tensors
- Rashaping - reshapes an input tensor to a defined shape
- View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
- Stacking - combine multiple tenosors on top of each other (cstack) or side by side (hstack)
- Squeeze - removes all 1 dimensions from a tensor
- Unsqueere - add a 1 dimension to a target tensor
- Permute - Return a view of the input with dimensions permuted (swapped) in a certain way



```python
# Let s create a tensor
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add extra dimension

x_reshape = x.reshape(9, 1)   # take a tensor of size [9] (dim = 1) and add one dimension to transform it to a [9, 1] (dim = 2)
x_reshape, x_reshape.shape, x_reshape.ndim
```

```python
x_reshape = x.reshape(1, 9)
x_reshape, x_reshape.shape
```

```python
# change the view
z = x.view(1, 9)
z, z.shape
```

```python
# Changing z changes x because a view of a tensor shares the same memory as the original
z [:, 0] = 5      # change position 0 in tensor from 1 to 5
z, x
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked
```

```python
# torch.squeeze() - eremoves all signle dimension from a target tensor
print(f"Previous tensor: {x_reshape}")
print(f"Previous shape: {x_reshape.shape}")



# remove extra dimensions from x_reshaped
x_squeezed = x_reshape.squeeze()        # squeeze removes all empty dimensions from a tensor
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")


print("  ")
print("  ")
x_reshape.shape
```

```python
# torch.unsqueeze() - adds a single dimension to a target tensor at a specific dim
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

# Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")

x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearrange
x_original = torch.rand(size=(224, 224, 3)) # height, width, colour
x_original.shape
print(f"Previous shape: {x_original.shape}")

# Permute the original tensor to rearrange the axis (or dim) order

x_permute = x_original.permute(2, 0, 1)     # shifts axis 0->1, 1->2, 2->0
x_permute.shape
print(f"New shape: {x_permute.shape}")      # shape (3, 224, 224)

# (2, 0, 1) are the dimensions
```

**Reproducibility**

Taking the random out of random

A neural network learns by starting with random numers, perform tensor operatons and update numbers, again, again, again

To be able to reproduce experiments, you don't want to have it too random.

To reduce the randomness, in PyTorch **random seed**.

RS = flavour the randomness


```python
import torch

# Create 2 random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(" ")
print(random_tensor_B)
print(" ")
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make random but reproducible tensors
import torch

# Set the random seed but it needs to be called at each block of code
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(" ")
print(random_tensor_D)
print(" ")
print(random_tensor_C == random_tensor_D)
```
